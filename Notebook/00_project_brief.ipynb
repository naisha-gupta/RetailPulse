# RetailPulse — Project Brief

## 1. Company Background

Olist is a Brazilian retail e-commerce company with 99,441 customers 
and 99,441 orders placed between 2016 and 2018. It connects small 
businesses and sellers across Brazil to customers through a single 
online marketplace, operating across 27 states with 3,095 sellers 
and 32,951 unique products.

## 2. Business Problem

Olist is facing three critical business challenges:

1. **Low repeat purchase rate** — 99,441 customers placed 99,441 orders,
   meaning almost every customer bought only once. Repeat purchasing 
   is nearly zero, which limits long term revenue growth.

2. **High delivery times** — average delivery takes 12 days, which is 
   significantly higher than global e-commerce standards.

3. **Geographic concentration** — 42% of all customers are from São Paulo 
   alone. 14+ states are severely underserved, representing untapped 
   market opportunity.

## 3. Business Questions

This project will answer the following business questions:

1. Which states and cities generate the most revenue?
2. What is driving the near-zero repeat purchase rate?
3. Which product categories generate the most revenue?
4. How does delivery time affect customer review scores?
5. Which products and categories receive the highest reviews?
6. Which sellers are the top performers on the platform?

## 4. Success Criteria

This project will be considered successful if it:

1. Answers all 6 business questions with data-backed evidence
2. Identifies at least 3 actionable business recommendations
3. Quantifies the impact of each recommendation where possible
4. Delivers an interactive dashboard accessible to non-technical stakeholders
5. Documents all findings in a clear, professional format

## 5. Constraints and Assumptions

### Data Constraints:
1. **Limited time period** — data covers only 2016 to 2018 (2 years).
   Long term trends and patterns cannot be fully established.

2. **No marketing data** — advertising spend and campaign data is 
   unavailable. We cannot calculate true customer acquisition cost 
   or measure marketing ROI.

3. **Brazil only** — analysis is limited to Brazilian market. 
   Findings may not apply to other geographies.

4. **No inventory data** — we cannot analyze stock levels or 
   supply chain performance beyond delivery times.

### Assumptions:
- customer_unique_id is the true identifier for repeat purchase analysis
- Delivered orders only will be used for revenue calculations
- Cancelled and unavailable orders will be excluded from KPI calculations

# Phase 8 — Business Insights and Recommendations

---

## Recommendation 1: Fix the Repeat Purchase Crisis

**Finding:**
Only 3.12% of customers ever place a second order.
96.88% of customers buy once and never return.
This is the single most critical business problem in the dataset.

**Root Cause:**
Customers in non-SP states wait 20+ days for delivery.
Our analysis proved that 1-star reviews average 20.8 delivery days
while 5-star reviews average only 10.2 days.
Poor delivery experience kills the motivation to reorder.

**Recommendation:**
1. Partner with regional logistics providers in Northern and
   Northeastern states to reduce delivery from 29 days to under 12
2. Build regional warehouses or fulfilment centres in RJ and MG
   to serve the second and third largest customer bases faster
3. Launch a post-purchase email campaign targeting one-time buyers
   offering a discount on their second order within 30 days

**Expected Impact:**
Increasing repeat rate from 3.12% to just 8% would double
revenue without acquiring a single new customer.

## Recommendation 2: Reduce Geographic Concentration Risk

**Finding:**
São Paulo generates 44% of total revenue despite Brazil
having 27 states. Top 3 states (SP, RJ, MG) contribute
74% of all revenue. 14+ states have fewer than 500 orders each.

**Root Cause:**
Most Olist sellers are based in São Paulo, making local
delivery fast and cheap. Customers in remote states face
long delivery times and high freight costs — reducing
their motivation to order online.

**Recommendation:**
1. Launch targeted digital marketing campaigns in RS, SC,
   and BA — states showing strong growth signals
2. Partner with local sellers in underserved states to
   reduce delivery times and freight costs
3. Offer free shipping promotions in states with fewer
   than 1,000 orders to stimulate first-time purchases

**Expected Impact:**
Our data shows customers spend similarly regardless of state.
Acquiring 10,000 new customers in underserved states at
the same AOV of BRL 154 would generate BRL 1.54M in new revenue.
This reduces SP dependency from 44% toward the 30% target.

## Recommendation 3: Reduce Delivery Time to Drive Review Scores

**Finding:**
A strong negative correlation exists between delivery time
and customer satisfaction:
- 1-star reviews → average 20.8 days delivery
- 5-star reviews → average 10.2 days delivery
Every additional day of delivery reduces review score.

**Root Cause:**
Olist relies heavily on São Paulo based logistics networks.
Customers in Northern and Northeastern states face 25-29 day
deliveries due to poor last-mile logistics infrastructure.
This directly causes 1-star reviews and kills repeat purchases.

**Recommendation:**
1. Set a maximum delivery time SLA of 10 days for all states
2. Partner with local sellers in underserved states to reduce
   last-mile delivery distance
3. Penalise sellers who consistently exceed delivery deadlines
4. Invest in logistics partnerships with regional carriers in
   RR, AM, AP and AC — the four worst performing states

**Expected Impact:**
Reducing average delivery from 12 days to 8 days would shift
average review score from 4.09 toward the 4.5 target.
Higher reviews directly drive repeat purchases — solving
Recommendation 1 simultaneously.

## Recommendation 4: Audit High-Revenue, High-Risk Categories

**Finding:**
health_beauty is the #2 revenue category at BRL 1.66M
but carries the highest product data quality risk.
610 products across the platform have missing category
names and incomplete descriptions — concentrated in
health, beauty and electronics categories.

**Root Cause:**
Sellers on the Olist marketplace self-upload product
listings with minimal quality checks. Incomplete or
misleading product descriptions lead to customer
disappointment, returns, and 1-star reviews.

**Recommendation:**
1. Implement mandatory product listing standards for
   health_beauty and computers_accessories categories
   — minimum photo count, description length, and
   accurate attribute fields (size, shade, weight)
2. Add a quality review step before new health_beauty
   products go live on the platform
3. Flag sellers with high return rates in these categories
   for performance review
4. Replace 610 null category products with correct
   classifications to improve search and discovery

**Expected Impact:**
Improving product listing quality in top 3 categories
reduces return rates, improves review scores, and
protects the BRL 4.93M revenue these categories generate.
A 5% reduction in returns saves approximately
BRL 246,000 in logistics and refund costs annually.
