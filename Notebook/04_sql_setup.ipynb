```python
# ============================================================
# RetailPulse | Phase 5: SQL Setup
# Push cleaned CSVs into MySQL
# ============================================================

import pandas as pd
from sqlalchemy import create_engine

# Connect to MySQL
engine = create_engine('mysql+pymysql://root:naisha25@localhost/retailpulse')

# Load your cleaned tables
PROCESSED = "../data/processed/"
RAW = "../data/raw/"

tables = {
    "customers"  : pd.read_csv(RAW + "olist_customers_dataset.csv"),
    "orders"     : pd.read_csv(PROCESSED + "orders_clean.csv"),
    "order_items": pd.read_csv(RAW + "olist_order_items_dataset.csv"),
    "payments"   : pd.read_csv(PROCESSED + "payments_clean.csv"),
    "products"   : pd.read_csv(PROCESSED + "products_clean.csv"),
    "sellers"    : pd.read_csv(RAW + "olist_sellers_dataset.csv"),
    "reviews"    : pd.read_csv(RAW + "olist_order_reviews_dataset.csv"),
    "categories" : pd.read_csv(RAW + "product_category_name_translation.csv")
}

# Push each table to MySQL
for name, df in tables.items():
    df.to_sql(name, engine, if_exists='replace', index=False)
    print(f"✓ Uploaded {name} → {len(df):,} rows")

print("\n✅ All tables loaded into MySQL!")
```
